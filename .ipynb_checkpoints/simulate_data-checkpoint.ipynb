```python
import pandas as pd
import numpy as np
```

```python
n = 4000
```

```python
rnd = np.random.RandomState(23404)

wallets = ['Blockchain', 'Electrum', 'Coinbase', 'Circle']

df = pd.DataFrame(index=range(n))
df['uid'] = df.index
df['walletx2'] = list(rnd.choice(wallets, size=n))

df['max_priv_intermediary'] = df.walletx2.isin(['Blockchain', 'Electrum']).astype(int)
df['max_priv_gov'] = df.walletx2.isin(['Electrum']).astype(int)
df['max_priv_public'] = df.walletx2.isin(['Coinbase', 'Circle']).astype(int)
df['banklike'] = df.walletx2.isin(['Coinbase', 'Circle']).astype(int)

df['ir_studyr'] = rnd.choice([1, 2, 3, 4], size=n)
binary_cols = ['male', 'topcoder', 'expectpricedown', 'open_source_browser', 'am_privacy_peers', 'am_privacy_int',
               'am_privacy_gov', 'trust_gov', 'trust_startup', 'trust_retailer', 'es_pub', 'es_int', 'es_gov',
               'rpgpextra', 'sb_cashing_out_btc_05', 'rmatrix']
for c in binary_cols:
    df[c] = np.random.choice([0, 1], size=n)

df['fw_notracePEERS'] = np.random.choice([1, 2, 3, 4, 5], size=n)
df['fw_notraceISP'] = np.random.choice([1, 2, 3, 4, 5], size=n)
df['fw_notraceGOV'] = np.random.choice([1, 2, 3, 4, 5], size=n)

df['wallet_order'] = [rnd.permutation(wallets) for _ in range(n)]
df['first_wallet'] = df.wallet_order.apply(lambda v: v[0])
for ix in range(1, 5):
    for wallet in wallets:
        df['r{0}{1}'.format(wallet.lower(), ix)] = df.wallet_order.apply(lambda v: int(wallet == v[ix - 1]))
# intermediaries (i.e. wallet providers) are Electrum and Blockchain, which do not fully share transaction data with the wallet provider. 
#The fully open sourced wallet Electrum maximizes privacy from the government. 
#Finally, the bank-like wallets Circle and Coinbase maximize privacy from the public given how they pool transactions.}

df['best_priv_public_n1st'] = (~df.first_wallet.isin(['Circle', 'Coinbase'])).astype(int)
df['best_priv_int_n1st'] = (~df.first_wallet.isin(['Electrum', 'Blockchain'])).astype(int)
df['best_priv_gov_n1st'] = (~df.first_wallet.isin(['Electrum'])).astype(int)
    # 'Escaping Surveillance from Public': 'es_pub',
    # 'Escaping Surveillance from Intermediary': 'es_int',
    # 'Escaping Surveillance from Government': 'es_gov'



df.to_csv('data_simulated.csv', sep=',', index=False)
df
```
